```python
# Part 1: Deriving the Marginal Density
import sympy as sp
from IPython.display import display, Markdown

# Use raw strings 'r' to prevent invalid escape sequence warnings
x_i, C_i, k, K, phi_k = sp.symbols(r'x_i C_i k K \phi_k')
Sigma_k, mu_k = sp.symbols(r'\Sigma_k \mu_k')

display(Markdown("## Part 1: Deriving the Marginal Density"))

# Defined as a raw multi-line string r"""...""" to ignore LaTeX backslashes
derivation_text = r"""
### **Mathematical Proof**
To find the marginal density $p(x_i)$, we apply the **Law of Total Probability** by summing the joint probability density $p(x_i, C_i = k)$ over all possible states of the latent cluster variable $C_i \in \{1, \dots, K\}$:

$$p(x_i) = \sum_{k=1}^{K} p(x_i, C_i = k)$$

Using the definition of conditional probability, we can write the joint density as:
$$p(x_i, C_i = k) = P(C_i = k) \cdot p(x_i \mid C_i = k)$$

Substituting the given model assumptions:
* Prior probability of cluster membership: $P(C_i = k) = \phi_k$
* Conditional observation density: $p(x_i \mid C_i = k) = \mathcal{N}(x_i \mid \mu_k, \Sigma_k)$

We obtain:
$$p(x_i) = \sum_{k=1}^{K} \phi_k \mathcal{N}(x_i \mid \mu_k, \Sigma_k)$$

---

### **Why is this density called a Gaussian Mixture Density?**
This density is called a **Gaussian Mixture Density** because it represents a weighted linear combination (or "mixture") of $K$ distinct component Gaussian probability distributions.
* The weights $\phi_k$ are the mixing coefficients representing the prior probability of belonging to each component. They satisfy $\phi_k \ge 0$ and $\sum_{k=1}^K \phi_k = 1$.
* Each component $\mathcal{N}(x_i \mid \mu_k, \Sigma_k)$ is a standard multivariate Gaussian distribution with its own specific parameters (mean $\mu_k$ and covariance matrix $\Sigma_k$).
"""

display(Markdown(derivation_text))
```

```python
# Part 2: Deriving the Posterior Cluster Probability
from IPython.display import display, Markdown

display(Markdown("## Part 2: Deriving the Posterior Cluster Probability"))

# Defined as a raw multi-line string (r""") to bypass Python's escape sequence warnings
derivation_text = r"""
### **Mathematical Proof**
For a fixed observation $x_i$, we wish to compute the posterior probability that the observation belongs to cluster $k$. Applying **Bayes' Rule**:

$$P(C_i = k \mid X_i = x_i) = \frac{p(X_i = x_i \mid C_i = k)P(C_i = k)}{p(x_i)}$$

By substituting the Law of Total Probability expansion for the marginal density $p(x_i)$ in the denominator:
$$p(x_i) = \sum_{j=1}^{K} p(X_i = x_i \mid C_i = j)P(C_i = j)$$

We establish the initial Bayes' relation:
$$P(C_i = k \mid X_i = x_i) = \frac{p(X_i = x_i \mid C_i = k)P(C_i = k)}{\sum_{j=1}^{K} p(X_i = x_i \mid C_i = j)P(C_i = j)}$$

Now, substituting our specific Gaussian model and cluster priors:
* $P(C_i = k) = \phi_k$
* $p(X_i = x_i \mid C_i = k) = \mathcal{N}(x_i \mid \mu_k, \Sigma_k)$

We get:
$$P(C_i = k \mid X_i = x_i) = \frac{\phi_k \mathcal{N}(x_i \mid \mu_k, \Sigma_k)}{\sum_{j=1}^{K} \phi_j \mathcal{N}(x_i \mid \mu_j, \Sigma_j)}$$

This quantity is defined as the **responsibility** of cluster $k$ for data point $x_i$, and is denoted by:
$$\gamma_{ik} = P(C_i = k \mid X_i = x_i)$$

---

### **Why may $\gamma_{ik}$ be interpreted as a posterior probability of cluster membership?**
The variable $\gamma_{ik}$ is a posterior probability because:
1. **Conditioning on Evidence**: It represents the updated belief about which cluster $C_i$ generated the data point $x_i$ *after* observing the actual data values of $x_i$.
2. **Probability Constraints**: Since it is derived directly from Bayes' theorem, it naturally satisfies:
   * $\gamma_{ik} \ge 0$ for all $k$.
   * $\sum_{k=1}^{K} \gamma_{ik} = 1$ for any given observation $x_i$.
"""

display(Markdown(derivation_text))
```

```python
# Part 3: One-Hot Encoding of the Latent Cluster Variable
from IPython.display import display, Markdown

display(Markdown("## Part 3: One-Hot Encoding of the Latent Cluster Variable"))

derivation_text = r"""
### **Mathematical Proof**
Let the latent random variable vector be $Z_i = [Z_{i1}, Z_{i2}, \dots, Z_{iK}]^T$ where:
$$Z_{ik} = \begin{cases} 1, & \text{if } C_i = k \\ 0, & \text{otherwise} \end{cases}$$

We wish to compute the conditional expectation of $Z_{ik}$ given $X_i = x_i$. By definition of the expectation of a discrete indicator variable:
$$\mathbb{E}[Z_{ik} \mid X_i = x_i] = \sum_{z \in \{0, 1\}} z \cdot P(Z_{ik} = z \mid X_i = x_i)$$

$$\mathbb{E}[Z_{ik} \mid X_i = x_i] = 0 \cdot P(Z_{ik} = 0 \mid X_i = x_i) + 1 \cdot P(Z_{ik} = 1 \mid X_i = x_i)$$

$$\mathbb{E}[Z_{ik} \mid X_i = x_i] = P(Z_{ik} = 1 \mid X_i = x_i)$$

Since $Z_{ik} = 1$ is logically equivalent to the event $C_i = k$, we have:
$$P(Z_{ik} = 1 \mid X_i = x_i) = P(C_i = k \mid X_i = x_i)$$

Hence:
$$\mathbb{E}[Z_{ik} \mid X_i = x_i] = P(C_i = k \mid X_i = x_i) = \gamma_{ik}$$

Applying this element-wise to the vector $Z_i$:
$$\mathbb{E}[Z_i \mid X_i = x_i] = \mathbb{E}\left[ \begin{bmatrix} Z_{i1} \\ Z_{i2} \\ \vdots \\ Z_{iK} \end{bmatrix} \Bigg| X_i = x_i \right] = \begin{bmatrix} \mathbb{E}[Z_{i1} \mid X_i = x_i] \\ \mathbb{E}[Z_{i2} \mid X_i = x_i] \\ \vdots \\ \mathbb{E}[Z_{iK} \mid X_i = x_i] \end{bmatrix} = \begin{bmatrix} \gamma_{i1} \\ \gamma_{i2} \\ \vdots \\ \gamma_{iK} \end{bmatrix}$$

### **Conclusion**
The soft cluster assignment in a Gaussian mixture model is precisely the conditional expectation:
$$\mathbb{E}[Z_i \mid X_i = x_i]$$
"""

display(Markdown(derivation_text))
```

```python
# Part 4: From Soft Assignment to Hard Clustering
from IPython.display import display, Markdown

display(Markdown("## Part 4: From Soft Assignment to Hard Clustering"))

explanation_text = r"""
### **Difference Between Soft Clustering and Hard Clustering**

| Feature | Soft Clustering | Hard Clustering |
| :--- | :--- | :--- |
| **Concept** | Assigns a fractional probability distribution of belonging to each of the $K$ clusters. | Assigns each data point $x_i$ to exactly one cluster. |
| **Mathematical Representation** | Represented by a continuous vector of expectations: <br> $\mathbb{E}[Z_i \mid X_i = x_i] = [\gamma_{i1}, \gamma_{i2}, \dots, \gamma_{iK}]^T$ where $\sum_k \gamma_{ik} = 1$. | Represented by a single discrete index: <br> $\widehat{C}_i = \arg\max_{1 \le k \le K} \gamma_{ik}$. |
| **Ambiguity Handling** | Retains uncertainty. A point near a cluster boundary might have responsibilities like $[0.48, 0.52, 0.0]$. | Discards uncertainty. The boundary point is forced into a single cluster $[0, 1, 0]$. |
| **Typical Algorithms** | Gaussian Mixture Model (GMM) posterior probability mapping. | K-Means, or GMM with maximum a posteriori (MAP) assignment. |
"""

display(Markdown(explanation_text))
```

```python
# Part 5: Conditional Expectation of the Observation Given the Cluster
from IPython.display import display, Markdown

display(Markdown("## Part 5: Conditional Expectation of the Observation Given the Cluster"))

derivation_text = r"""
### **Mathematical Proof**
Under the model formulation, conditional on the event that $C_i = k$, the observation vector $X_i$ is distributed according to a multivariate Gaussian distribution:
$$X_i \mid C_i = k \sim \mathcal{N}(\mu_k, \Sigma_k)$$

By the fundamental definition of the mean (expected value) of a multivariate Gaussian distribution:
$$\mathbb{E}[X_i \mid C_i = k] = \mu_k$$

---

### **Why $\mu_k$ can be interpreted as the center of cluster $k$**
The mean vector $\mu_k$ acts as the first moment (the balance point) of the probability distribution representing cluster $k$. Because the density is symmetric and unimodal (Gaussian), $\mu_k$ represents the peak density location, pointing to the exact spatial center of that cluster in $\mathbb{R}^d$.

---

### **Comparison of the Two Conditional Expectations**

1. **Expectation 1: $\mathbb{E}[Z_i \mid X_i = x_i]$**
   * **Domain & Output**: Maps from $\mathbb{R}^d$ to a simplex in $\mathbb{R}^K$.
   * **Interpretation**: This represents **soft cluster membership**. Given an observed point $x_i$, it outputs the posterior probabilities of which cluster generated it.

2. **Expectation 2: $\mathbb{E}[X_i \mid C_i = k]$**
   * **Domain & Output**: Maps from cluster label $k \in \{1, \dots, K\}$ to $\mathbb{R}^d$.
   * **Interpretation**: This represents the **cluster mean location**. Assuming we are examining cluster $k$, it outputs the expected value of the generated physical data point.
"""

display(Markdown(derivation_text))
```

```python
# Part 6: The Complete-Data Likelihood
from IPython.display import display, Markdown

display(Markdown("## Part 6: The Complete-Data Likelihood"))

derivation_text = r"""
### **Mathematical Derivation**
If the latent variables $z_i$ were known, the complete-data likelihood for $n$ independent observations is:
$$p(x_1, \dots, x_n, z_1, \dots, z_n) = \prod_{i=1}^{n} \prod_{k=1}^{K} \left[ \phi_k \mathcal{N}(x_i \mid \mu_k, \Sigma_k) \right]^{z_{ik}}$$

Taking the natural logarithm ($\ln$ or $\ell_c$) of both sides:
$$\ell_c = \ln p(x_1, \dots, x_n, z_1, \dots, z_n)$$

$$\ell_c = \ln \left( \prod_{i=1}^{n} \prod_{k=1}^{K} \left[ \phi_k \mathcal{N}(x_i \mid \mu_k, \Sigma_k) \right]^{z_{ik}} \right)$$

Using logarithmic properties ($\ln(ab) = \ln a + \ln b$ and $\ln(a^b) = b \ln a$):
$$\ell_c = \sum_{i=1}^{n} \sum_{k=1}^{K} z_{ik} \ln \left[ \phi_k \mathcal{N}(x_i \mid \mu_k, \Sigma_k) \right]$$

$$\ell_c = \sum_{i=1}^{n} \sum_{k=1}^{K} z_{ik} \left[ \ln \phi_k + \ln \mathcal{N}(x_i \mid \mu_k, \Sigma_k) \right]$$

---

### **Why would this expression be easy to maximize if $z_{ik}$ were known?**
If the indicator variables $z_{ik}$ were observed, maximization becomes trivial because:
1. **Decoupling of Components**: The double summation splits the parameters of the $K$ different clusters. Instead of dealing with a complex log-of-sum (from the marginal mixture), we have a sum-of-logs.
2. **Independent Sub-problems**: The optimization of each $(\mu_k, \Sigma_k)$ can be solved independently by focusing only on the data points that belong to cluster $k$ (where $z_{ik} = 1$). This reduces the optimization to standard, closed-form Maximum Likelihood Estimation (MLE) of individual Gaussians.
"""

display(Markdown(derivation_text))
```

```python
# Part 7: The EM Interpretation
from IPython.display import display, Markdown

display(Markdown("## Part 7: The EM Interpretation"))

derivation_text = r"""
### **Mathematical Formulation**
In practice, the latent indicators $z_{ik}$ are unobserved. The Expectation-Maximization (EM) algorithm handles this by taking the conditional expectation of the complete-data log-likelihood with respect to the posterior distribution of the latent variables, conditioned on the observed data and current parameters:

$$Q = \mathbb{E}_{Z \mid X} [\ell_c]$$

Substituting the complete-data log-likelihood expression:
$$Q = \mathbb{E}_{Z \mid X} \left[ \sum_{i=1}^{n} \sum_{k=1}^{K} Z_{ik} \left( \ln \phi_k + \ln \mathcal{N}(x_i \mid \mu_k, \Sigma_k) \right) \right]$$

Using the linearity of expectation:
$$Q = \sum_{i=1}^{n} \sum_{k=1}^{K} \mathbb{E}[Z_{ik} \mid X_i = x_i] \left[ \ln \phi_k + \ln \mathcal{N}(x_i \mid \mu_k, \Sigma_k) \right]$$

Since we proved in Part 3 that $\mathbb{E}[Z_{ik} \mid X_i = x_i] = \gamma_{ik}$, we substitute this to get the expected complete-data log-likelihood:
$$Q = \sum_{i=1}^{n} \sum_{k=1}^{K} \gamma_{ik} \left[ \ln \phi_k + \ln \mathcal{N}(x_i \mid \mu_k, \Sigma_k) \right]$$

---

### **Why the E-step is a conditional update of cluster membership probabilities**
The E-step computes the current posterior responsibilities $\gamma_{ik}$ using the current parameter estimates $(\phi_k, \mu_k, \Sigma_k)$. This updates our probabilistic assignment (fractional memberships) for every observation $x_i$ based on how likely each Gaussian component is to have generated it, given our latest parameter settings.
"""

display(Markdown(derivation_text))
```

```python
# Part 8: Parameter Updates (M-Step Derivation)
from IPython.display import display, Markdown

display(Markdown("## Part 8: Parameter Updates"))

derivation_text = r"""
### **Mathematical Derivations**

To update the parameters, we maximize $Q$ with respect to $\phi_k$, $\mu_k$, and $\Sigma_k$.

#### **1. Updating the Mean $\mu_k$**
We take the partial derivative of $Q$ with respect to $\mu_k$:
$$\frac{\partial Q}{\partial \mu_k} = \frac{\partial}{\partial \mu_k} \sum_{i=1}^{n} \gamma_{ik} \ln \mathcal{N}(x_i \mid \mu_k, \Sigma_k)$$

Recall the log-density of a Gaussian:
$$\ln \mathcal{N}(x_i \mid \mu_k, \Sigma_k) = -\frac{d}{2}\ln(2\pi) - \frac{1}{2}\ln|\Sigma_k| - \frac{1}{2}(x_i - \mu_k)^T \Sigma_k^{-1} (x_i - \mu_k)$$

Taking the derivative:
$$\frac{\partial Q}{\partial \mu_k} = \sum_{i=1}^{n} \gamma_{ik} \Sigma_k^{-1} (x_i - \mu_k) = 0$$

Multiplying by $\Sigma_k$:
$$\sum_{i=1}^{n} \gamma_{ik} x_i - \sum_{i=1}^{n} \gamma_{ik} \mu_k = 0 \implies \mu_k^{\text{new}} = \frac{\sum_{i=1}^{n} \gamma_{ik} x_i}{\sum_{i=1}^{n} \gamma_{ik}} = \frac{1}{N_k}\sum_{i=1}^{n} \gamma_{ik} x_i$$
where $N_k = \sum_{i=1}^{n} \gamma_{ik}$.

#### **2. Updating the Covariance $\Sigma_k$**
Similarly, maximizing $Q$ with respect to $\Sigma_k$ yields:
$$\Sigma_k^{\text{new}} = \frac{1}{N_k} \sum_{i=1}^{n} \gamma_{ik} (x_i - \mu_k^{\text{new}})(x_i - \mu_k^{\text{new}})^T$$

#### **3. Updating the Mixture Weights $\phi_k$**
We maximize $Q$ under the constraint that $\sum_{k=1}^{K} \phi_k = 1$ using a Lagrange multiplier $\lambda$:
$$\mathcal{L}(\Phi, \lambda) = Q + \lambda \left( \sum_{k=1}^{K} \phi_k - 1 \right)$$

Taking the derivative with respect to $\phi_k$:
$$\frac{\partial \mathcal{L}}{\partial \phi_k} = \sum_{i=1}^{n} \frac{\gamma_{ik}}{\phi_k} + \lambda = 0 \implies \phi_k = -\frac{1}{\lambda} \sum_{i=1}^{n} \gamma_{ik} = -\frac{N_k}{\lambda}$$

Summing over all $k$:
$$\sum_{k=1}^{K} \phi_k = -\frac{1}{\lambda} \sum_{k=1}^{K} N_k \implies 1 = -\frac{n}{\lambda} \implies \lambda = -n$$

Substituting $\lambda$ back:
$$\phi_k^{\text{new}} = \frac{N_k}{n}$$

---

### **How $\gamma_{ik}$ acts as a fractional membership weight**
In these equations, $\gamma_{ik}$ represents a continuous value between $0$ and $1$. It acts as a fractional weight showing how much of data point $x_i$ belongs to cluster $k$.
* Instead of data point $x_i$ only contributing to one cluster (as in hard clustering where the weights are strictly $0$ or $1$), it contributes fractionally to all clusters based on its responsibility values.
"""

display(Markdown(derivation_text))
```

```python
# Part 9: Interpretation of GMM as Conditional Updating
from IPython.display import display, Markdown

display(Markdown("## Part 9: Interpretation of GMM as Conditional Updating"))

interpretation_text = r"""
### **Summary of GMM as a Process of Conditional Updating**

Gaussian Mixture Model (GMM) clustering can be viewed as an iterative, two-stage conditional updating cycle:

1. **Prior Beliefs**:
   The mixture weight $\phi_k$ represents the prior probability $P(C_i = k)$ that any randomly selected point belongs to cluster $k$ before we look at its features.

2. **Compatibility Assessment**:
   The Gaussian density $\mathcal{N}(x_i \mid \mu_k, \Sigma_k)$ acts as a likelihood function. It measures how compatible the physical location of observation $x_i$ is with the profile of cluster $k$.

3. **Bayesian Posterior Update**:
   By combining the prior and likelihood using Bayes' rule, we calculate the responsibility:
   $$\gamma_{ik} = P(C_i = k \mid X_i = x_i)$$
   This represents the updated, conditional soft assignment vector $\mathbb{E}[Z_i \mid X_i = x_i]$.

4. **Parameter Realignment (The M-Step)**:
   The parameters $(\phi_k, \mu_k, \Sigma_k)$ are updated using these posterior probabilities as weights. This dynamically realigns the clusters to better represent the data points assigned to them.

This continuous cycle of updating assignments given parameters (E-step) and updating parameters given assignments (M-step) makes GMM a robust, probabilistic clustering technique built entirely on conditional expectations.
"""

display(Markdown(interpretation_text))
```

```python
# Part 10: Computational Simulation and Out-of-Sample Validation
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from sklearn.preprocessing import StandardScaler
from sklearn.mixture import GaussianMixture
from sklearn.model_selection import train_test_split
from IPython.display import display, Markdown

display(Markdown("## Part 10: Computational Simulation & Out-of-Sample Validation"))

# 1. Load sample dataset mimicking the Kaggle credit card dataset structure
np.random.seed(42)
n_samples = 1000

# Cluster 1: Low Purchases, Low Credit Limit
c1 = np.random.multivariate_normal([1000, 3000], [[400000, 100000], [100000, 900000]], int(n_samples*0.4))
# Cluster 2: High Purchases, High Credit Limit
c2 = np.random.multivariate_normal([8000, 15000], [[2000000, 1000000], [1000000, 5000000]], int(n_samples*0.3))
# Cluster 3: Moderate Purchases, Medium-High Credit Limit
c3 = np.random.multivariate_normal([3500, 9000], [[900000, 200000], [200000, 1600000]], int(n_samples*0.3))

data = np.vstack([c1, c2, c3])
df = pd.DataFrame(data, columns=['PURCHASES', 'CREDIT_LIMIT'])

# 2. Data Splitting and Scaling
X = df[['PURCHASES', 'CREDIT_LIMIT']].values
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test = train_test_split(X_scaled, test_size=0.20, random_state=42)

print(f"Dataset Split: {X_train.shape[0]} Training samples, {X_test.shape[0]} Test samples.")

# 3. EM Execution
gmm = GaussianMixture(n_components=3, covariance_type='full', random_state=42)
gmm.fit(X_train)

print(f"Model Converged: {gmm.converged_}")
print(f"Iterations Required: {gmm.n_iter_}")

# 4. Out-of-Sample Performance
test_score = gmm.score(X_test) # returns average log-likelihood
print(f"Average Log-Likelihood on Test Set: {test_score:.4f}\n")

# --- Interactive Visualizations ---

# Generate coordinate grid for contours
x_min, x_max = X_scaled[:, 0].min() - 0.5, X_scaled[:, 0].max() + 0.5
y_min, y_max = X_scaled[:, 1].min() - 0.5, X_scaled[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200), np.linspace(y_min, y_max, 200))
grid_points = np.c_[xx.ravel(), yy.ravel()]

# Calculate probabilities across the coordinate grid
probs = gmm.predict_proba(grid_points)
max_probs = probs.max(axis=1).reshape(xx.shape)

# 1. 2D Density Heatmap of raw training data
fig1 = px.density_heatmap(
    x=X_train[:, 0], y=X_train[:, 1],
    marginal_x="histogram", marginal_y="histogram",
    labels={'x': 'Scaled PURCHASES', 'y': 'Scaled CREDIT_LIMIT'},
    title="1. Empirical 2D Density Heatmap (Training Data)"
)
fig1.show()

# 2. Training Assignment Plot with Contour Map
fig2 = go.Figure()
# Contour showing maximum posterior responsibility
fig2.add_trace(go.Contour(
    x=np.linspace(x_min, x_max, 200),
    y=np.linspace(y_min, y_max, 200),
    z=max_probs,
    colorscale='Viridis',
    contours_coloring='heatmap',
    name='Max Responsibility Contour',
    colorbar=dict(title='Posterior Prob')
))
# Scatter plot of actual training points
train_preds = gmm.predict(X_train)
fig2.add_trace(go.Scatter(
    x=X_train[:, 0], y=X_train[:, 1],
    mode='markers',
    marker=dict(color=train_preds, colorscale='Viridis', line=dict(width=1, color='white'), size=6),
    name='Train Points'
))
fig2.update_layout(
    title="2. Training Assignment Plot overlaid on Contour Map of Maximum Responsibility",
    xaxis_title="Scaled PURCHASES", yaxis_title="Scaled CREDIT_LIMIT"
)
fig2.show()

# 3. Test Assignment Plot (Out-of-sample overlay)
fig3 = go.Figure()
fig3.add_trace(go.Contour(
    x=np.linspace(x_min, x_max, 200),
    y=np.linspace(y_min, y_max, 200),
    z=max_probs,
    colorscale='Viridis',
    contours_coloring='heatmap',
    name='Max Responsibility Contour',
    colorbar=dict(title='Posterior Prob')
))
# Scatter plot of test points
test_preds = gmm.predict(X_test)
fig3.add_trace(go.Scatter(
    x=X_test[:, 0], y=X_test[:, 1],
    mode='markers',
    marker=dict(color=test_preds, colorscale='Plasma', line=dict(width=1, color='black'), size=8),
    name='Test Points'
))
fig3.update_layout(
    title="3. Test Assignment Plot (Out-of-Sample) overlaid on Contour Map",
    xaxis_title="Scaled PURCHASES", yaxis_title="Scaled CREDIT_LIMIT"
)
fig3.show()

# Evaluation write-up
# Using raw string (r) here to avoid syntax warnings on LaTeX mathematical symbols
evaluation_text = r"""
### **Evaluation of Visualizations and Theoretical Alignment**

1. **Empirical 2D Density Heatmap**:
   This visualization clearly shows multiple peaks (multimodality), confirming that a single standard Gaussian distribution would be insufficient. A Gaussian Mixture Model is well-suited here.

2. **Training and Test Assignment Overlays**:
   The background contour map shows the maximum posterior responsibility $\max_k \gamma_{ik}$ computed across the feature space:
   * **Deep Yellow/High Value Regions**: These represent high-certainty areas where the maximum posterior responsibility is close to $1.0$. Here, points belong unambiguously to a single cluster.
   * **Purple/Low Value Boundaries**: These represent transition regions where the maximum responsibility drops close to $1/K = 0.33$. This directly visualizes the soft assignment vector $\mathbb{E}[Z_i \mid X_i = x_{\text{grid}}]$ we derived in **Part 3**. Instead of showing hard, abrupt boundaries, the gradual color transitions represent the continuous, probabilistic changes in cluster membership.
"""
display(Markdown(evaluation_text))
```
