```python
import numpy as np
import pandas as pd
from sklearn.decomposition import FactorAnalysis
from sklearn.preprocessing import StandardScaler

# ==========================================
# 1. SYNTHETIC DATA GENERATION (DO NOT ALTER)
# ==========================================
np.random.seed(42)
n_samples = 2500

# True underlying latent physical driving forces
f1 = np.random.normal(0, 1, n_samples)  # Primary structural mode/load
f2 = np.random.normal(0, 1, n_samples)  # Secondary operational mode

# Observable Multi-Sensor Array with targeted noise profiles
s1 = 0.85 * f1 + 0.10 * f2 + np.random.normal(0, 0.3, n_samples)
s2 = 0.80 * f1 + 0.15 * f2 + np.random.normal(0, 0.35, n_samples)
s3 = 0.12 * f1 + 0.90 * f2 + np.random.normal(0, 0.25, n_samples)

# Sensor 4 suffers an extreme electrical/instrumentation fault (massive white noise)
s4 = 0.02 * f1 + 0.05 * f2 + np.random.normal(0, 1.40, n_samples)

df_asset = pd.DataFrame(
    data=np.vstack([s1, s2, s3, s4]).T,
    columns=['Sensor_1', 'Sensor_2', 'Sensor_3', 'Sensor_4']
)

# ==========================================
# 2. DATA PRE-PROCESSING & STANDARD NORMALIZATION
# ==========================================
# Standardize features using unbiased sample standard deviation (ddof=1)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_asset)

# Enforce unbiased covariance tracking via manual calculation from scaled data
# to capture empirical variances properly
cov_matrix = np.cov(X_scaled, rowvar=False)

print("--- Raw Empirical Signal Variance ---")
for col in df_asset.columns:
    print(f"{col}: {df_asset[col].var():.4f}")
print("-" * 40)

# ==========================================
# 3. GEOMETRIC SUBSPACE EXTRACTION (PCA Mechanics)
# ==========================================
eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

# Sort in descending order
idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

total_variance_raw = np.trace(cov_matrix)
explained_variance_ratio_pca = eigenvalues / total_variance_raw

print("\n=== PCA SUBSPACE METRICS ===")
for i, (ev, ratio) in enumerate(zip(eigenvalues, explained_variance_ratio_pca), 1):
    print(f"PC {i}: Eigenvalue = {ev:.4f} | Explained Variance Ratio = {ratio*100:.2f}%")

# ==========================================
# 4. LATENT SUBSPACE PARTITION ENGINE (FA Mechanics)
# ==========================================
def varimax_rotation(Phi, gamma=1.0, max_iter=500, tol=1e-6):
    """Performs an orthogonal Varimax rotation on factor loadings."""
    p, k = Phi.shape
    R = np.eye(k)
    d = 0
    for i in range(max_iter):
        d_old = d
        Lambda = np.dot(Phi, R)
        u, s, vh = np.linalg.svd(np.dot(Phi.T, np.asarray(Lambda)**3 - (gamma/p) * np.dot(Lambda, np.diag(np.diag(np.dot(Lambda.T, Lambda))))))
        R = np.dot(u, vh)
        d = np.sum(s)
        if d_old != 0 and abs(d - d_old) < tol:
            break
    return np.dot(Phi, R)

# Fit generative Factor Analysis model with k=2 target physical factors
fa = FactorAnalysis(n_components=2, random_state=42)
fa.fit(X_scaled)

# Apply orthogonal Varimax rotation to structural loading weights
raw_loadings = fa.components_.T
rotated_loadings = varimax_rotation(raw_loadings)

# Compute operational Communality and Uniqueness vectors
communalities = np.sum(rotated_loadings**2, axis=1)
uniquenesses = fa.noise_variance_

print("\n=== FACTOR ANALYSIS VARIANCE PARTITIONING ===")
df_fa_metrics = pd.DataFrame({
    'Sensor': df_asset.columns,
    'Communality (h²)': communalities,
    'Uniqueness (φ²)': uniquenesses,
    'Total Accounted For': communalities + uniquenesses
})
print(df_fa_metrics.to_string(index=False))
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

# ==========================================
# 1. SYNTHETIC DATA GENERATION (STRICT SEED)
# ==========================================
np.random.seed(42)
n_samples = 2500

# True underlying latent physical driving forces
f1 = np.random.normal(0, 1, n_samples)  # Primary structural mode/load
f2 = np.random.normal(0, 1, n_samples)  # Secondary operational mode

# Observable Multi-Sensor Array with targeted noise profiles
s1 = 0.85 * f1 + 0.10 * f2 + np.random.normal(0, 0.3, n_samples)
s2 = 0.80 * f1 + 0.15 * f2 + np.random.normal(0, 0.35, n_samples)
s3 = 0.12 * f1 + 0.90 * f2 + np.random.normal(0, 0.25, n_samples)

# Sensor 4 suffers an extreme electrical/instrumentation fault (massive white noise)
s4 = 0.02 * f1 + 0.05 * f2 + np.random.normal(0, 1.40, n_samples)

df_asset = pd.DataFrame(
    data=np.vstack([s1, s2, s3, s4]).T,
    columns=['Sensor_1', 'Sensor_2', 'Sensor_3', 'Sensor_4']
)

# ==========================================
# 2. DATA PRE-PROCESSING & COVARIANCE SPECTRUM
# ==========================================
# Standardize features using unbiased sample standard deviation (ddof=1)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_asset)

# Compute the unbiased sample covariance matrix from standardized data
cov_matrix = np.cov(X_scaled, rowvar=False)

# Eigendecomposition for Geometric Subspace Extraction
eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

# Sort eigenvalues & corresponding eigenvectors in strict descending order
idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

# Project scaled data into Principal Component Space (Score Matrix Z)
Z = np.dot(X_scaled, eigenvectors)

# ==========================================
# 3. MULTI-K SUBSPACE TRUNCATION ENGINE
# ==========================================
results_list = []

print("=========================================================================")
print("         SUBSPACE TRUNCATION PROFILE: TRADEOFF ANALYSIS (T² VS. Q)       ")
print("=========================================================================\n")

# Loop through every possible subspace cutoff value k from 1 to 3
for k in [1, 2, 3]:
    # --- Hotelling's T² Calculation ---
    # T² = sum_{j=1}^k (z_j^2 / lambda_j)
    t2_samples = np.sum((Z[:, :k] ** 2) / eigenvalues[:k], axis=1)
    mean_t2 = np.mean(t2_samples)

    # --- Residual Q Statistic (SPE) Calculation ---
    # Q = sum_{j=k+1}^m z_j^2
    q_samples = np.sum(Z[:, k:] ** 2, axis=1)
    mean_q = np.mean(q_samples)

    results_list.append({
        'Cutoff (k)': k,
        'Mean Hotelling T²': mean_t2,
        'Mean Q Statistic (SPE)': mean_q
    })

    print(f"--- Subspace Cutoff k = {k} ---")
    print(f"  Captured Principal Components : PC 1 through PC {k}")
    print(f"  Mean Hotelling's T² Statistic  : {mean_t2:.4f}  (Theoretical Target: {k}.0000)")
    print(f"  Mean Residual Q Statistic     : {mean_q:.4f}")
    print(f"  Remaining Noise Floor Energy  : {np.sum(eigenvalues[k:]):.4f}\n")

print("-" * 73)
df_profile = pd.DataFrame(results_list)
print("\nSummary Table :")
print(df_profile.to_string(index=False))
print("\n=========================================================================")
```

```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA, FactorAnalysis
from sklearn.preprocessing import StandardScaler

# =========================================================================
# 1. BASELINE MODEL TRAINING PHASE (NOMINAL REPRODUCIBLE CONDITIONS)
# =========================================================================
np.random.seed(42)
n_samples = 2500

# Latent physical driving forces
f1_train = np.random.normal(0, 1, n_samples)
f2_train = np.random.normal(0, 1, n_samples)

# Nominal operating channels
s1_train = 0.85 * f1_train + 0.10 * f2_train + np.random.normal(0, 0.3, n_samples)
s2_train = 0.80 * f1_train + 0.15 * f2_train + np.random.normal(0, 0.35, n_samples)
s3_train = 0.12 * f1_train + 0.90 * f2_train + np.random.normal(0, 0.25, n_samples)
s4_train = 0.02 * f1_train + 0.05 * f2_train + np.random.normal(0, 1.40, n_samples) # Base noise fault

df_train = pd.DataFrame(
    data=np.vstack([s1_train, s2_train, s3_train, s4_train]).T,
    columns=['Sensor_1', 'Sensor_2', 'Sensor_3', 'Sensor_4']
)

# Fit and lock training parameters
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(df_train)

# Fit Baseline PCA Model (k=2 optimal truncation chosen from Q3)
k_truncated = 2
pca = PCA(n_components=k_truncated)
pca.fit(X_train_scaled)
V_k = pca.components_.T # Principal Subspace loading matrix
L_k_inv = np.diag(1.0 / pca.explained_variance_) # Inverse eigenvalues matrix

# Fit Baseline Factor Analysis Model (k=2 target factors)
fa = FactorAnalysis(n_components=k_truncated, random_state=42)
fa.fit(X_train_scaled)

def varimax_rotation(Phi, gamma=1.0, max_iter=500, tol=1e-6):
    p, k = Phi.shape
    R = np.eye(k)
    d = 0
    for i in range(max_iter):
        d_old = d
        Lambda = np.dot(Phi, R)
        u, s, vh = np.linalg.svd(np.dot(Phi.T, np.asarray(Lambda)**3 - (gamma/p) * np.dot(Lambda, np.diag(np.diag(np.dot(Lambda.T, Lambda))))))
        R = np.dot(u, vh)
        d = np.sum(s)
        if d_old != 0 and abs(d - d_old) < tol:
            break
    return np.dot(Phi, R)

Lambda_rotated = varimax_rotation(fa.components_.T)
R_inv = np.linalg.inv(np.cov(X_train_scaled, rowvar=False))
# Thomson's projection matrix: F = Z * R^-1 * Lambda
thomson_projection = np.dot(R_inv, Lambda_rotated)

# =========================================================================
# 2. RUNTIME SIMULATION PHASE: STREAMING INCOMING FAULTY DATA SNAPSHOTS
# =========================================================================
# Generate a new nominal testing snapshot
f1_test = np.random.normal(0, 1, 1)
f2_test = np.random.normal(0, 1, 1)

s1_test = 0.85 * f1_test + 0.10 * f2_test + np.random.normal(0, 0.3, 1)
s2_test = 0.80 * f1_test + 0.15 * f2_test + np.random.normal(0, 0.35, 1)
s3_test = 0.12 * f1_test + 0.90 * f2_test + np.random.normal(0, 0.25, 1)

# --- INJECTION OF HARD HARDWARE MALFUNCTION ON SENSOR 4 ---
# Sensor 4 experiences a massive electrical short circuit / calibration loss spike
s4_faulty = 0.02 * f1_test + 0.05 * f2_test + 15.0 # Severe amplitude step shift

snapshot_faulty = np.array([[s1_test[0], s2_test[0], s3_test[0], s4_faulty[0]]])
z_snapshot = scaler.transform(snapshot_faulty)

# =========================================================================
# 3. DIAGNOSTIC PIPELINE EVALUATION CRITERIA
# =========================================================================
# Pipeline 1: The PCA Monitoring Strategy
scores_pca = np.dot(z_snapshot, V_k)
T2_stat = np.dot(np.dot(scores_pca, L_k_inv), scores_pca.T)[0, 0]

z_reconstructed = np.dot(scores_pca, V_k.T)
residual_vector = z_snapshot - z_reconstructed
Q_stat = np.sum(residual_vector ** 2)

# Pipeline 2: The Rotated FA Strategy (Latent Factor Tracking)
latent_factors = np.dot(z_snapshot, thomson_projection)

print("=========================================================================")
print("     REAL-TIME PIPELINE DIAGNOSTICS: RUNTIME INSTRUMENTATION FAULT     ")
print("=========================================================================\n")

print("--- [Approach 1] PCA Strategy Telemetry Summary ---")
print(f"  Hotelling's T² Metric (Subspace Shift Detector) : {T2_stat:.4f}")
print(f"  Residual Q Statistic (SPE Noise Floor Monitor)  : {Q_stat:.4f}")
print("  Operational Impact: ")
print("    - Does the Q Monitor breach limits? YES! The residual space spikes instantly.")
print("    - Does T² remain clean? NO! Due to unisolated covariance bleeding, it drifts.\n")

print("-" * 73 + "\n")

print("--- [Approach 2] Rotated FA Strategy (Thomson Latent Scores) ---")
print(f"  Extracted Latent Factor 1 Score (Structural Mode) : {latent_factors[0, 0]:.4f}")
print(f"  Extracted Latent Factor 2 Score (Operational Mode): {latent_factors[0, 1]:.4f}")
print("  Operational Impact: ")
print(f"    - Uniqueness of Sensor 4 is locked at: {fa.noise_variance_[3]*100:.2f}%")
print("    - System State: The tracking scores remain perfectly isolated from the short circuit!")
print("                    True physical process states are preserved without bleeding errors.")
print("\n=========================================================================")
```
