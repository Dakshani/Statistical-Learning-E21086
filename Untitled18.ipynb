```python
# ==============================================================================
# BAYESIAN ESTIMATION OF STRUCTURAL STIFFNESS EFFICIENCY (THETA)
# Clean code using updated NumPy trapezoid integration
# ==============================================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta, norm

%matplotlib inline
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['figure.dpi'] = 120

# ------------------------------------------------------------------------------
# PARAMETERS & DATA SETUP
# ------------------------------------------------------------------------------
theta_true = 0.75
K_nominal = 100.0
sigma_noise = 0.15

alpha_prior = 8.0
beta_prior = 1.5

M = 500
theta_grid = np.linspace(0.01, 1.0, M)

np.random.seed(42)
num_steps = 5
noise_realizations = np.random.normal(0, sigma_noise, num_steps)
y_obs = theta_true * K_nominal * np.exp(noise_realizations)

# ------------------------------------------------------------------------------
# TASK 1: PRIOR
# ------------------------------------------------------------------------------
prior_pdf = beta.pdf(theta_grid, alpha_prior, beta_prior)
analytical_prior_mean = alpha_prior / (alpha_prior + beta_prior)

print(f"TASK 1 SUMMARY:")
print(f"Analytical Expected Prior Stiffness Efficiency E[theta^(0)]: {analytical_prior_mean:.4f} ({analytical_prior_mean*100:.2f}%)")

plt.figure(figsize=(8, 4.5))
plt.plot(theta_grid, prior_pdf, color='#1f77b4', linewidth=2.5, label=f'Prior Beta({alpha_prior}, {beta_prior})')
plt.axvline(x=analytical_prior_mean, color='#ff7f0e', linestyle='--', linewidth=2,
            label=f'Expected Mean E[θ] = {analytical_prior_mean:.4f}')
plt.title("Task 1: Initial Prior Probability Density Function", fontsize=12, fontweight='bold')
plt.xlabel("Stiffness Efficiency Parameter (θ)", fontsize=10)
plt.ylabel("Probability Density f(θ)", fontsize=10)
plt.xlim([0, 1.05])
plt.ylim(bottom=0)
plt.legend(loc='upper left', frameon=True)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------------------------
# TASKS 2-6: LIKELIHOOD & UPDATES (Using np.trapezoid to avoid warnings)
# ------------------------------------------------------------------------------
def log_normal_likelihood(y, theta_vec, K_nom, sigma):
    mu_ln = np.log(theta_vec * K_nom)
    return (1.0 / (y * sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((np.log(y) - mu_ln) / sigma)**2)

posteriors = [prior_pdf.copy()]
current_posterior = prior_pdf.copy()

print("\nSEQUENTIAL BAYESIAN UPDATES:")
for k in range(num_steps):
    y_k = y_obs[k]
    likelihood_k = log_normal_likelihood(y_k, theta_grid, K_nominal, sigma_noise)
    unnormalized_posterior = likelihood_k * current_posterior

    # Updated to np.trapezoid (or np.trapz if running older numpy)
    if hasattr(np, 'trapezoid'):
        normalizing_constant = np.trapezoid(unnormalized_posterior, theta_grid)
        current_posterior = unnormalized_posterior / normalizing_constant
        bayes_mean_k = np.trapezoid(theta_grid * current_posterior, theta_grid)
    else:
        normalizing_constant = np.trapz(unnormalized_posterior, theta_grid)
        current_posterior = unnormalized_posterior / normalizing_constant
        bayes_mean_k = np.trapz(theta_grid * current_posterior, theta_grid)

    posteriors.append(current_posterior.copy())
    print(f"Step k={k+1}: Measurement y_{k+1}={y_k:.2f} | Estimated E[θ|y] = {bayes_mean_k:.4f}")

plt.figure(figsize=(9, 5))
plt.plot(theta_grid, posteriors[0], label='Prior (k=0)', color='gray', linestyle=':', linewidth=2)
colors = ['#1f77b4', '#2ca02c', '#d62728', '#9467bd', '#8c564b']
for k in range(1, num_steps + 1):
    plt.plot(theta_grid, posteriors[k], label=f'Posterior k={k}', color=colors[k-1], linewidth=2)

plt.axvline(x=theta_true, color='black', linestyle='--', linewidth=2, label=f'True θ = {theta_true}')
plt.title("Task 6: Sequential Bayes Posterior Probability Distributions", fontsize=12, fontweight='bold')
plt.xlabel("Stiffness Efficiency Parameter (θ)", fontsize=10)
plt.ylabel("Probability Density f(θ | y^(k))", fontsize=10)
plt.xlim([0.2, 1.0])
plt.ylim(bottom=0)
plt.legend(loc='upper left', frameon=True)
plt.tight_layout()
plt.show()
```
