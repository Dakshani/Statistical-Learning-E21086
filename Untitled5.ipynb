```python
import numpy as np

def solve_shooting_target(n_samples=1000000):
    print(f"Running simulation with {n_samples:,} samples...\n")

    # --- Scenario 1: Uniformly distributed with respect to area ---
    # To get uniform area, we use r = R * sqrt(U) where U is Uniform(0,1)
    u = np.random.uniform(0, 1, n_samples)
    r1 = 10 * np.sqrt(u)
    p_a1 = np.mean(r1 > 9)

    # --- Scenario 2: Distance r is Uniform(0, 10) ---
    r2 = np.random.uniform(0, 10, n_samples)
    p_a2 = np.mean(r2 > 9)

    # Theoretical Calculations for verification
    # Theory 1: (Area of disk) - (Area of inner disk) / Total Area
    theory1 = (np.pi * 10**2 - np.pi * 9**2) / (np.pi * 10**2)
    # Theory 2: Length of interval [9, 10] / Length of interval [0, 10]
    theory2 = (10 - 9) / 10

    print("--- Results ---")
    print(f"1. Uniform Area Distribution:")
    print(f"   Simulated P(A): {p_a1:.4f}")
    print(f"   Theoretical P(A): {theory1:.4f} (19%)\n")

    print(f"2. Uniform Distance Distribution:")
    print(f"   Simulated P(A): {p_a2:.4f}")
    print(f"   Theoretical P(A): {theory2:.4f} (10%)\n")

    print("--- Explanation ---")
    print("The answers differ because the 'density' of points is distributed differently.")
    print("In Model 1, the probability is proportional to the area. Since the outer ring")
    print("has more area than a ring of the same thickness near the center, P(A) is higher.")
    print("In Model 2, every 'thickness' of 1 unit has the same probability regardless of")
    print("how much area it actually covers, leading to a linear (and lower) probability.")

solve_shooting_target()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def simulate_bertrand_rich(n_chords=200):
    fig, axes = plt.subplots(1, 3, figsize=(20, 7))
    titles = ['1. Random Endpoints (P=1/3)',
              '2. Random Radial Dist (P=1/2)',
              '3. Random Midpoint (P=1/4)']

    # Pre-calculating the critical distance (R/2)
    # Chord length > side of inscribed triangle if midpoint distance < 0.5
    threshold = 0.5

    for i in range(3):
        ax = axes[i]
        ax.set_aspect('equal')
        # Draw the main circle and the threshold circle
        circle = plt.Circle((0, 0), 1, color='black', fill=False, lw=2)
        inner = plt.Circle((0, 0), 0.5, color='red', fill=False, linestyle='--', alpha=0.5)
        ax.add_artist(circle)
        ax.add_artist(inner)

        successes = 0

        for _ in range(n_chords):
            if i == 0: # Method 1: Random Endpoints
                theta1, theta2 = np.random.uniform(0, 2*np.pi, 2)
                p1 = [np.cos(theta1), np.sin(theta1)]
                p2 = [np.cos(theta2), np.sin(theta2)]
                dist_from_center = np.abs(np.cos((theta1 - theta2) / 2))

            elif i == 1: # Method 2: Random Radial Distance
                theta = np.random.uniform(0, 2*np.pi)
                r = np.random.uniform(0, 1)
                # Find midpoint at distance r, then find perpendicular chord
                midpoint = np.array([r * np.cos(theta), r * np.sin(theta)])
                dist_from_center = r
                # Calculate endpoints using chord half-length
                h = np.sqrt(max(0, 1 - r**2))
                perp_dir = np.array([-np.sin(theta), np.cos(theta)])
                p1, p2 = midpoint + h*perp_dir, midpoint - h*perp_dir

            else: # Method 3: Random Midpoint (Uniform Area)
                r = np.sqrt(np.random.uniform(0, 1))
                theta = np.random.uniform(0, 2*np.pi)
                midpoint = np.array([r * np.cos(theta), r * np.sin(theta)])
                dist_from_center = r
                h = np.sqrt(max(0, 1 - r**2))
                perp_dir = np.array([-np.sin(theta), np.cos(theta)])
                p1, p2 = midpoint + h*perp_dir, midpoint - h*perp_dir

            # Determine color: Blue if longer than triangle side, Gray otherwise
            is_long = dist_from_center < threshold
            if is_long: successes += 1
            color = 'royalblue' if is_long else 'lightgray'
            ax.plot([p1[0], p2[0]], [p1[1], p2[1]], color=color, lw=0.8, alpha=0.6)

        ax.set_title(f"{titles[i]}\nSimulated P ≈ {successes/n_chords:.3f}")
        ax.axis('off')

    plt.suptitle("Bertrand's Paradox: Visualizing Chord Distributions", fontsize=16)
    plt.show()

simulate_bertrand_rich()
```

```python
import pandas as pd
import numpy as np

def simulate_boy_girl_paradox(n_families=100000):
    # 1. Generate a large number of families with two children
    # 0 = Boy, 1 = Girl
    child1 = np.random.randint(0, 2, n_families)
    child2 = np.random.randint(0, 2, n_families)

    df = pd.DataFrame({'Child1': child1, 'Child2': child2})

    # Define labels for clarity
    gender = {0: 'Boy', 1: 'Girl'}
    df['Label'] = df['Child1'].map(gender) + "-" + df['Child2'].map(gender)

    # 2. Apply the condition: "at least one is a girl"
    # This means Child1 is a girl OR Child2 is a girl
    at_least_one_girl = df[(df['Child1'] == 1) | (df['Child2'] == 1)]

    # 3. Calculate how many of these have "both girls"
    both_girls = at_least_one_girl[(at_least_one_girl['Child1'] == 1) &
                                   (at_least_one_girl['Child2'] == 1)]

    prob = len(both_girls) / len(at_least_one_girl)

    print(f"Total families simulated: {n_families:,}")
    print(f"Families with at least one girl: {len(at_least_one_girl):,}")
    print(f"Families with two girls: {len(both_girls):,}")
    print(f"\nSimulated Probability: {prob:.4f}")
    print(f"Theoretical Probability: {1/3:.4f} (1/3)")

    # Display the sample space breakdown
    print("\nSample Space Breakdown (Given at least one girl):")
    print(at_least_one_girl['Label'].value_counts())

simulate_boy_girl_paradox()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from itertools import chain, combinations

def get_powerset(iterable):
    s = list(iterable)
    return list(chain.from_iterable(combinations(s, r) for r in range(len(s)+1)))

def solve_q9_complete():
    # --- 1. SETUP ---
    omega = {'G1', 'G2', 'D1', 'D2'}
    probs = {'G1': 0.50, 'G2': 0.20, 'D1': 0.10, 'D2': 0.20}

    # Mapping for Random Variables
    X_map = {'G1': 0, 'G2': 0, 'D1': 1, 'D2': 1}
    Y_map = {'G1': 1, 'G2': 2, 'D1': 3, 'D2': 4}

    print("=== Q9: FORMAL PROBABILITY SPACE ANSWERS ===\n")

    # Part 1: Probability Space
    f_sigma = get_powerset(omega)
    print(f"1. Probability Space (Ω, F, P):")
    print(f"   Ω = {omega}")
    print(f"   F (Power Set) contains {len(f_sigma)} subsets.")
    print(f"   P values: {probs}\n")

    # Part 2 & 3: Sigma Algebras
    sigma_x = [set(), {'G1', 'G2'}, {'D1', 'D2'}, omega]
    sigma_y = [set(s) for s in f_sigma]
    print(f"2. σ(X) = {sigma_x}")
    print(f"3. σ(Y) = Power Set of Ω (contains all 16 subsets)\n")

    # Part 4: Information Inclusion
    print(f"4. Is σ(X) ⊆ σ(Y)? {'Yes' if all(s in sigma_y for s in sigma_x) else 'No'}")
    print("   Reason: Y is an injective mapping; it preserves all distinctions in Ω.\n")

    # Part 5 & 6: Marginals
    p_x0 = probs['G1'] + probs['G2']
    p_x1 = probs['D1'] + probs['D2']
    print(f"5. Marginal X: P(X=0) = {p_x0:.2f}, P(X=1) = {p_x1:.2f}")
    print(f"6. Marginal Y: P(Y=1)={probs['G1']}, P(Y=2)={probs['G2']}, P(Y=3)={probs['D1']}, P(Y=4)={probs['D2']}\n")

    # Part 7: Joint
    print(f"7. Joint Probabilities:")
    print(f"   P(X=0, Y=1) = {probs['G1']} | P(X=0, Y=2) = {probs['G2']}")
    print(f"   P(X=1, Y=3) = {probs['D1']} | P(X=1, Y=4) = {probs['D2']}\n")

    # Part 8: Conditionals
    print(f"8. Conditional Probabilities:")
    print(f"   P(X=1 | Y=3) = {probs['D1']/probs['D1']:.2f}")
    print(f"   P(Y=3 | X=1) = {probs['D1']/p_x1:.3f}")
    print(f"   P(Y=4 | X=1) = {probs['D2']/p_x1:.3f}\n")

    # Part 9 & 10: Interpretations
    print("9.  Conditioning on X provides 'coarse' info (Quality).")
    print("    Conditioning on Y provides 'fine' info (Specific Machine).")
    print("10. σ(Y) contains Machine Identity (e.g., event {G1}) which σ(X) cannot see.\n")

    # --- 11. VISUALIZATION ---
    fig, ax = plt.subplots(1, 2, figsize=(14, 5))

    # Chart X
    ax[0].bar(['Good (X=0)', 'Defect (X=1)'], [p_x0, p_x1], color=['#2ecc71', '#e74c3c'])
    ax[0].set_title("Information in σ(X)")
    ax[0].set_ylabel("Probability")

    # Chart Y
    ax[1].bar(['G1', 'G2', 'D1', 'D2'], probs.values(), color=['#27ae60', '#2ecc71', '#c0392b', '#e74c3c'])
    ax[1].set_title("Information in σ(Y)")

    plt.tight_layout()
    plt.show()

solve_q9_complete()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

def solve_q10_continuous_full():
    print("=== Q10: INFORMATION IN CONTINUOUS RANDOM VARIABLES ===\n")

    # 1. Probability Space
    print("1. Probability Space: (Ω, F, P) where Ω = R², F = B(R²), and P is the measure from joint density f_TP.")

    # 2 & 3. Sigma-Algebras
    print("2. σ(X) = { B × R : B ∈ B(R) }. This represents all vertical 'strips' in the plane.")
    print("3. σ(Y) = B(R²). This is the full Borel σ-algebra of the plane.\n")

    # 5 & 6. Marginals
    print("5. Marginal Density of X: f_X(t) = (1/√(2π)) * exp(-t²/2)  [Standard Normal N(0,1)]")
    print("6. Marginal Density of Y: f_Y(t,p) = f_T(t) * f_P(p) [Multivariate Normal]\n")

    # 7. Numerical Probability Calculations
    p_x_le_0 = stats.norm.cdf(0, 0, 1)
    p_x_gt_1 = 1 - stats.norm.cdf(1, 0, 1)
    # P(Y in [-1,1]x[-2,2]) = P(-1<T<1) * P(-2<P<2)
    p_y_rect = (stats.norm.cdf(1, 0, 1) - stats.norm.cdf(-1, 0, 1)) * \
               (stats.norm.cdf(2, 0, 2) - stats.norm.cdf(-2, 0, 2))

    print(f"7. Computed Probabilities:")
    print(f"   P(X <= 0) = {p_x_le_0:.4f}")
    print(f"   P(X > 1)  = {p_x_gt_1:.4f}")
    print(f"   P(Y ∈ [-1,1] × [-2,2]) = {p_y_rect:.4f}\n")

    # 12. VISUALIZATIONS
    fig = plt.figure(figsize=(18, 5))

    # A. Joint Density Heat Map (for Y)
    t = np.linspace(-4, 4, 100)
    p = np.linspace(-6, 6, 100)
    T, P = np.meshgrid(t, p)
    Z = (1/(4*np.pi)) * np.exp(-(T**2/2 + P**2/8))

    ax1 = fig.add_subplot(1, 3, 1)
    cp = ax1.contourf(T, P, Z, cmap='magma')
    fig.colorbar(cp, ax=ax1)
    ax1.set_title("Joint Density f_Y(t,p)\n(Heat Map)")
    ax1.set_xlabel("Temperature (t)")
    ax1.set_ylabel("Pressure (p)")

    # B. Marginal Density of X
    ax2 = fig.add_subplot(1, 3, 2)
    ax2.plot(t, stats.norm.pdf(t, 0, 1), color='blue', lw=2)
    ax2.fill_between(t, stats.norm.pdf(t, 0, 1), color='blue', alpha=0.2)
    ax2.set_title("Marginal Density of X\n(Standard Normal)")
    ax2.set_xlabel("t")

    # C. Coarseness Diagram (sigma(X) vs sigma(Y))
    ax3 = fig.add_subplot(1, 3, 3)
    # sigma(X) can only see sets that are vertical strips
    for x_val in [-2, -1, 0, 1, 2]:
        ax3.axvline(x_val, color='blue', alpha=0.3, linestyle='--')
    ax3.fill_between([0.5, 1.5], -6, 6, color='blue', alpha=0.1, label='Set in σ(X) (Strip)')
    # sigma(Y) can see specific points or small regions
    ax3.add_patch(plt.Rectangle((0.7, 2), 0.5, 1, color='red', alpha=0.6, label='Set in σ(Y) (Box)'))

    ax3.set_xlim(-4, 4); ax3.set_ylim(-6, 6)
    ax3.set_title("σ(X) is Coarser than σ(Y)\n(Strips vs Localized Regions)")
    ax3.legend(loc='upper right', fontsize='small')

    plt.tight_layout()
    plt.show()

solve_q10_continuous_full()
```
