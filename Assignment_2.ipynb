```python
import numpy as np
import matplotlib.pyplot as plt

# --- Data Setup ---
x_values = np.array([0, 1, 2, 3])
probabilities = np.array([0.50, 0.30, 0.15, 0.05])

# 1 & 2. Definitions (Conceptual for the code)
print("--- Probability Space & Random Variable ---")
print(f"Sample Space (Omega): {x_values}")
print(f"Probabilities (P): {probabilities}")
print(f"Constraint Check (Sum = 1): {np.sum(probabilities)}\n")

# 3. Probability Mass Function (PMF)
def pmf(k):
    if k in x_values:
        return probabilities[np.where(x_values == k)[0][0]]
    return 0

print("--- 3. Probability Mass Function ---")
for val in x_values:
    print(f"P(X = {val}) = {pmf(val)}")
print("")

# 4. Compute Expectation E[X]
# Formula: E[X] = sum(x * P(x))
expectation = np.sum(x_values * probabilities)
print(f"--- 4. Expectation ---")
print(f"E[X] = {expectation:.2f} failures per day\n")

# 5. Compute Variance Var(X)
# Formula: Var(X) = E[X^2] - (E[X])^2
expectation_x2 = np.sum((x_values**2) * probabilities)
variance = expectation_x2 - (expectation**2)
print(f"--- 5. Variance ---")
print(f"Var(X) = {variance:.4f}\n")

# 6. Find P(X >= 1) and P(X >= 2)
p_ge_1 = np.sum(probabilities[x_values >= 1])
p_ge_2 = np.sum(probabilities[x_values >= 2])
print(f"--- 6. Cumulative Probabilities ---")
print(f"P(X >= 1) = {p_ge_1:.2f}")
print(f"P(X >= 2) = {p_ge_2:.2f}\n")

# 7. Practical Meaning
print("--- 7. Practical Meaning ---")
print(f"Expectation ({expectation:.2f}): The long-term average number of failures the district should expect daily.")
print(f"Variance ({variance:.4f}): A measure of how much the daily failure count fluctuates or deviates from that average.")

# Visualization
plt.bar(x_values, probabilities, color='skyblue', edgecolor='navy')
plt.title('PMF of Transformer Failures')
plt.xlabel('Number of Failures (X)')
plt.ylabel('Probability P(X=x)')
plt.xticks(x_values)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
import numpy as np
import scipy.stats as stats
from scipy.special import beta as beta_func
import matplotlib.pyplot as plt

# --- Configuration ---
alpha, beta_val = 2, 18

# 1. Probability Space
print("1. PROBABILITY SPACE (Omega, F, P)")
print("Sample Space (Omega): {x \u2208 R | 0 \u2264 x \u2264 1}")
print("Sigma-algebra (F): Borel sigma-algebra B([0, 1])")
print("Probability Measure (P): P(A) = \u222B_A f(x) dx")
print("-" * 30)

# 2. Random Variable Definition
print("2. RANDOM VARIABLE X")
print("X: \u03a9 \u2192 [0, 1] is a continuous random variable representing")
print("the proportion of defective items in a batch.")
print("-" * 30)

# 3. Probability Density Function (PDF)
print("3. PROBABILITY DENSITY FUNCTION")
print(f"f(x) = [1 / B({alpha}, {beta_val})] * x^({alpha}-1) * (1-x)^({beta_val}-1)")
print(f"f(x) = [1 / B(2, 18)] * x * (1-x)^17, for 0 < x < 1")
print("-" * 30)

# 4. Verification
b_const = beta_func(alpha, beta_val)
print("4. VERIFICATION OF PDF")
print(f"The constant B(2, 18) = {b_const:.6f}")
print("Verification: \u222B f(x) dx = [1/B(2,18)] * \u222B x(1-x)^17 dx")
print("Result: B(2,18) / B(2,18) = 1.0 (Confirmed)")
print("-" * 30)

# 5. Expectation
expectation = alpha / (alpha + beta_val)
print("5. EXPECTATION E[X]")
print(f"Formula: E[X] = \u03b1 / (\u03b1 + \u03b2) = 2 / (2 + 18)")
print(f"E[X] = {expectation:.2f}")
print("-" * 30)

# 6. Variance
variance = (alpha * beta_val) / ((alpha + beta_val)**2 * (alpha + beta_val + 1))
print("6. VARIANCE Var(X)")
print(f"Formula: Var(X) = (\u03b1\u03b2) / [(\u03b1+\u03b2)\u00b2 * (\u03b1+\u03b2+1)]")
print(f"Var(X) = 36 / (400 * 21) = {variance:.6f}")
print("-" * 30)

# 7. Probability P(X <= 0.10)
p7 = stats.beta.cdf(0.10, alpha, beta_val)
print("7. P(X \u2264 0.10)")
print("Integral Form: \u222B_{0}^{0.10} [1/B(2,18)] * x * (1-x)^17 dx")
print(f"Numerical Value: {p7:.4f}")
print("-" * 30)

# 8. Probability P(0.05 <= X <= 0.15)
p8 = stats.beta.cdf(0.15, alpha, beta_val) - stats.beta.cdf(0.05, alpha, beta_val)
print("8. P(0.05 \u2264 X \u2264 0.15)")
print("Integral Form: \u222B_{0.05}^{0.15} [1/B(2,18)] * x * (1-x)^17 dx")
print(f"Numerical Value: {p8:.4f}")
print("-" * 30)

# 9. Reasoning
print("9. WHY BETA DISTRIBUTION?")
print("1. Bounded Support: Proportions must exist in [0, 1].")
print("2. Flexibility: Models skewed data typical of defect rates.")
print("-" * 30)

# 10. Contextual Interpretation
print("10. QUALITY CONTEXT")
print(f"Expectation: The average daily defect rate is {expectation*100:.0f}%.")
print(f"Variance: Measures quality consistency; low variance = stable process.")

# Plotting
x = np.linspace(0, 0.5, 500)
plt.fill_between(x, stats.beta.pdf(x, alpha, beta_val), color='skyblue', alpha=0.4)
plt.plot(x, stats.beta.pdf(x, alpha, beta_val), color='blue', lw=2)
plt.title(f'Defect Proportion: Beta(\u03b1={alpha}, \u03b2={beta_val})')
plt.axvline(expectation, color='red', linestyle='--', label=f'Mean ({expectation})')
plt.xlabel('Proportion of Defects')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True, alpha=0.2)
plt.show()
```

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# --- Parameters ---
# a = lower limit, b = upper limit, c = mode
a, b, c = 4, 10, 6

def print_line():
    print("-" * 40)

# 1. Support of X
print("1. SUPPORT OF X")
print(f"The support is the interval [a, b], which is: [{a}, {b}] hours.")
print_line()

# 2. Probability Density Function (PDF)
print("2. PROBABILITY DENSITY FUNCTION f_X(x)")
print(f"For {a} \u2264 x < {c}: f(x) = 2(x - {a}) / [({b} - {a})({c} - {a})]")
print(f"For {c} \u2264 x \u2264 {b}: f(x) = 2({b} - x) / [({b} - {a})({b} - {c})]")
print("Otherwise: f(x) = 0")
print_line()

# 3. Verification
# Area of a triangle = 0.5 * base * height
base = b - a
height = 2 / (b - a)
area = 0.5 * base * height
print("3. VERIFICATION")
print(f"Height at mode (c): 2 / ({b} - {a}) = {height:.4f}")
print(f"Total Area = 0.5 * (base) * (height) = 0.5 * {base} * {height:.4f} = {area:.1f}")
print("Confirmed: Integrates to 1.")
print_line()

# 4. Expectation E[X]
# Formula: (a + b + c) / 3
expectation = (a + b + c) / 3
print("4. EXPECTATION E[X]")
print(f"E[X] = ({a} + {b} + {c}) / 3 = {expectation:.2f} hours")
print_line()

# 5. Variance Var(X)
# Formula: (a^2 + b^2 + c^2 - ab - ac - bc) / 18
variance = (a**2 + b**2 + c**2 - a*b - a*c - b*c) / 18
print("5. VARIANCE Var(X)")
print(f"Var(X) = {variance:.4f}")
print_line()

# 6. Context of a, b, c
print("6. CONTEXTUAL MEANING")
print(f"a ({a}h): Minimum time required for installation.")
print(f"b ({b}h): Maximum time required (worst-case scenario).")
print(f"c ({c}h): Most likely completion time (the peak probability).")
print_line()

# 7. Why Triangular?
print("7. REASONING")
print("It is used when data is scarce but expert judgment can provide")
print("the minimum, maximum, and most likely values.")
print_line()

# 8. P(X <= 5)
# Since 5 is between a and c, we use the first part of the CDF: (x-a)^2 / [(b-a)(c-a)]
prob_le_5 = ((5 - a)**2) / ((b - a) * (c - a))
print("8. P(X \u2264 5)")
print(f"P(X \u2264 5) = (5 - {a})\u00b2 / [({b} - {a})({c} - {a})] = {prob_le_5:.4f}")
print_line()

# 9. P(5 <= X <= 8)
# P(5 <= X <= 8) = P(X <= 8) - P(X <= 5)
# For x=8 (between c and b), CDF = 1 - (b-x)^2 / [(b-a)(b-c)]
prob_le_8 = 1 - ((b - 8)**2) / ((b - a) * (b - c))
prob_between = prob_le_8 - prob_le_5
print("9. P(5 \u2264 X \u2264 8)")
print(f"P(X \u2264 8) = {prob_le_8:.4f}")
print(f"P(5 \u2264 X \u2264 8) = {prob_le_8:.4f} - {prob_le_5:.4f} = {prob_between:.4f}")
print_line()

# 10. Planning & Sensitivity
print("10. SENSITIVITY ANALYSIS")
print("Helps identify risks: if variance is high, the timeline is uncertain.")
print("Used in Monte Carlo simulations to predict total project duration.")

# Plotting for visualization
x = np.linspace(a - 1, b + 1, 1000)
pdf = stats.triang.pdf(x, (c - a) / (b - a), loc=a, scale=b - a)
plt.plot(x, pdf, 'b-', lw=2, label='Triangular PDF')
plt.fill_between(x, pdf, where=((x >= 5) & (x <= 8)), color='orange', alpha=0.3, label='P(5 \u2264 X \u2264 8)')
plt.title(f'Project Completion Time: Triang({a}, {b}, {c})')
plt.xlabel('Hours')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
from scipy.special import beta as beta_func

# --- Parameters ---
a, b, c = 8, 20, 11

# 1. Compute alpha and beta shape parameters
alpha_param = 1 + 4 * ((c - a) / (b - a))
beta_param = 1 + 4 * ((b - c) / (b - a))

print(f"1. SHAPE PARAMETERS")
print(f"Alpha (\u03b1) = 1 + 4 * ({c}-{a})/({b}-{a}) = {alpha_param:.2f}")
print(f"Beta (\u03b2) = 1 + 4 * ({b}-{c})/({b}-{a}) = {beta_param:.2f}\n")

# 2. Support of X
print(f"2. SUPPORT OF X")
print(f"The support is the interval [a, b] = [{a}, {b}] days.\n")

# 3. PDF Function
print(f"3. PROBABILITY DENSITY FUNCTION f(x)")
print(f"f(x) = [(x - {a})^(\u03b1-1) * ({b} - x)^(\u03b2-1)] / [B(\u03b1, \u03b2) * ({b} - {a})^(\u03b1+\u03b2-1)]")
print(f"f(x) = [(x - 8)^{alpha_param-1:.1f} * (20 - x)^{beta_param-1:.1f}] / [B({alpha_param:.1f}, {beta_param:.1f}) * 12^{alpha_param+beta_param-1:.1f}]\n")

# 4 & 5. Probability Space and Random Variable
print(f"4. PROBABILITY SPACE (\u03a9, F, P)")
print(f"Sample Space (\u03a9): [8, 20]")
print(f"Sigma-algebra (F): Borel sigma-algebra B([8, 20])")
print(f"Probability Measure (P): P(A) = \u222B_A f(x) dx\n")

print(f"5. DEFINITION OF X")
print(f"X is a continuous random variable mapping \u03a9 to R representing task duration.\n")

# 6. Expectation E[X]
# Show that E[X] = (a + 4c + b) / 6
expected_val = (a + 4*c + b) / 6
print(f"6. EXPECTATION E[X]")
print(f"E[X] = ({a} + 4*{c} + {b}) / 6 = {expected_val:.2f} days\n")

# 7. Variance Var(X)
# Formula: Var(X) = [(E[X] - a)(b - E[X])] / 7
variance = ((expected_val - a) * (b - expected_val)) / 7
print(f"7. VARIANCE Var(X)")
print(f"Var(X) = [({expected_val:.2f} - {a}) * ({b} - {expected_val:.2f})] / 7 = {variance:.4f}\n")

# 8. Standard Deviation sigma
std_dev = np.sqrt(variance)
print(f"8. STANDARD DEVIATION (\u03c3)")
print(f"\u03c3 = \u221aVar(X) = \u221a{variance:.4f} = {std_dev:.4f} days\n")

# 9. PERT vs Triangular
# In a Triangular dist: E[X] = (a+b+c)/3 = (8+20+11)/3 = 13.0
tri_mean = (a + b + c) / 3
print(f"9. COMPARISON (PERT vs Triangular)")
print(f"Triangular Mean = {tri_mean:.2f}, PERT Mean = {expected_val:.2f}")
print("PERT is often better as it reduces the weight of extreme outliers (a and b)")
print("and provides a smoother, more realistic curve than the 'sharp' triangular peak.\n")

# 10. Practical Meaning
print(f"10. PRACTICAL MEANING")
print(f"a, b, c: The expert-defined best, worst, and most likely timeframes.")
print(f"E[X] ({expected_val:.2f}): The 'realistic' average completion date used for scheduling.")
print(f"\u03c3 ({std_dev:.4f}): The risk level; higher sigma means more uncertainty in the schedule.")

# Visualization
x = np.linspace(a, b, 500)
# PERT is a Beta distribution shifted and scaled
# stats.beta uses (alpha, beta, loc, scale)
y = stats.beta.pdf(x, alpha_param, beta_param, loc=a, scale=b-a)

plt.figure(figsize=(10, 5))
plt.plot(x, y, 'b-', lw=3, label='PERT Distribution')
plt.axvline(expected_val, color='r', linestyle='--', label=f'E[X] = {expected_val:.2f}')
plt.axvline(c, color='g', linestyle=':', label=f'Mode (c) = {c}')
plt.title(f'Software Integration Task: PERT({a}, {b}, {c})')
plt.xlabel('Days')
plt.ylabel('Density')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# --- Parameters ---
a = 5   # Minimum time
c = 11  # Most likely time

# 1. Determine the value of maximum time b
# Formula: E[X] = (a + 4c + b) / 6
# Given E[X] = c:
# c = (a + 4c + b) / 6
# 6c = a + 4c + b
# b = 2c - a

b = 2*c - a

print("1. MAXIMUM TIME (b)")
print(f"Given E[X] = c, then b = 2c - a")
print(f"b = 2({c}) - {a} = {b} days\n")

# 2. What happens to the variance?
# Formula: Var(X) = [(E[X] - a)(b - E[X])] / 7
# If E[X] = c, Var(X) = [(c - a)(b - c)] / 7

expected_val = c
variance = ((expected_val - a) * (b - expected_val)) / 7

print("2. VARIANCE BEHAVIOR")
print(f"With E[X] = {expected_val} and b = {b}:")
print(f"Var(X) = [({expected_val} - {a}) * ({b} - {expected_val})] / 7")
print(f"Var(X) = [({c-a}) * ({b-c})] / 7 = {variance:.4f}")
print("\nInsight: When E[X] matches the most likely time, the distribution")
print("becomes symmetric (a and b are equidistant from c).")
```
