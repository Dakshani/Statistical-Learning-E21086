```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import itertools

# ------------------------------------------------------------------------------
# 0. Setup and Data Frame Modeling of the Sample Space (Omega)
# ------------------------------------------------------------------------------
print("="*80)
print("INITIALIZING SAMPLE SPACE SETUP")
print("="*80)

# Define outcomes, sample space probabilities, and Random Variable mappings
omega = ['w1', 'w2', 'w3', 'w4']
probabilities = np.array([0.50, 0.10, 0.20, 0.20])
X_val = np.array([0, 1, 0, 1])  # Fraud Indicator
Y_val = np.array([0, 0, 1, 1])  # Risk Class Indicator

df = pd.DataFrame({
    'Outcome (omega)': omega,
    'Probability P(omega)': probabilities,
    'Fraud Indicator (X)': X_val,
    'Risk Class (Y)': Y_val
})
print(df.to_string(index=False))
print("\n" + "-"*80 + "\n")


# ------------------------------------------------------------------------------
# Q1. Write down the probability space (Omega, F, P)
# ------------------------------------------------------------------------------
print("QUESTION 1: Probability Space Structure")
print("-"*40)
print(f"Sample Space (Omega) = {set(omega)}")
print("Sigma-algebra (F) = Power Set P(Omega) containing 2^4 = 16 elements:")

# Generating power set
def get_powerset(s):
    return [set(x) for r in range(len(s) + 1) for x in itertools.combinations(s, r)]

f_sigma_algebra = get_powerset(omega)
for idx, s in enumerate(f_sigma_algebra):
    # Compute probability of each set in the sigma-algebra
    p_s = sum([probabilities[omega.index(elem)] for elem in s])
    if idx < 4 or idx > 12:  # Sample printing to save terminal space cleanly
        print(f"  Set {idx+1}: {s if s else '{}'} -> P = {p_s:.2f}")
    elif idx == 4:
        print("  ... [remaining subsets omitted for brevity] ...")


# ------------------------------------------------------------------------------
# Q2. Compute the sigma-algebra generated by Y, namely sigma(Y)
# ------------------------------------------------------------------------------
print("\nQUESTION 2: Sigma-Algebra generated by Y, sigma(Y)")
print("-"*40)
# Y groups {w1, w2} into 0, and {w3, w4} into 1.
set_Y0 = {'w1', 'w2'}
set_Y1 = {'w3', 'w4'}
sigma_Y = [set(), set_Y0, set_Y1, set(omega)]

print("sigma(Y) contains the following 4 events:")
for idx, s in enumerate(sigma_Y):
    print(f"  Element {idx+1}: {s if s else '{}'}")


# ------------------------------------------------------------------------------
# Q3. Explain why E[X|Y] must be sigma(Y)-measurable
# ------------------------------------------------------------------------------
print("\nQUESTION 3: Measure-Theoretic Rationalization of E[X|Y]")
print("-"*40)
explanation = (
    "By definition, the conditional expectation E[X|Y] is a random variable that\n"
    "is measurable with respect to the sub-sigma-algebra sigma(Y). This ensures that\n"
    "E[X|Y] does not contain any more granular information than what is provided by Y.\n"
    "Because the atoms (the finest non-empty disjoint sets) of sigma(Y) are exactly\n"
    "A0 = {w1, w2} (where Y=0) and A1 = {w3, w4} (where Y=1), any function that is\n"
    "sigma(Y)-measurable MUST be flat/constant over these specific sets of outcomes."
)
print(explanation)


# ------------------------------------------------------------------------------
# Q4. Compute E[X|Y] explicitly
# ------------------------------------------------------------------------------
print("\nQUESTION 4: Explicit Calculation of E[X|Y]")
print("-"*40)
# E[X | Y=0] = P(X=1, Y=0) / P(Y=0) = P(w2) / (P(w1) + P(w2))
p_y0 = df[df['Risk Class (Y)'] == 0]['Probability P(omega)'].sum()
p_x1_y0 = df[(df['Fraud Indicator (X)'] == 1) & (df['Risk Class (Y)'] == 0)]['Probability P(omega)'].sum()
E_X_Y0 = p_x1_y0 / p_y0

# E[X | Y=1] = P(X=1, Y=1) / P(Y=1) = P(w4) / (P(w3) + P(w4))
p_y1 = df[df['Risk Class (Y)'] == 1]['Probability P(omega)'].sum()
p_x1_y1 = df[(df['Fraud Indicator (X)'] == 1) & (df['Risk Class (Y)'] == 1)]['Probability P(omega)'].sum()
E_X_Y1 = p_x1_y1 / p_y1

print(f"P(Y=0) = {p_y0:.2f}, P(X=1, Y=0) = {p_x1_y0:.2f} => E[X|Y=0] = {E_X_Y0:.6f} (or 1/6)")
print(f"P(Y=1) = {p_y1:.2f}, P(X=1, Y=1) = {p_x1_y1:.2f} => E[X|Y=1] = {E_X_Y1:.6f} (or 1/2)")

# Vector mapping of E[X|Y] for each omega outcome
E_X_Y_vec = np.where(Y_val == 0, E_X_Y0, E_X_Y1)
df['E[X|Y]'] = E_X_Y_vec


# ------------------------------------------------------------------------------
# Q5. Verify the defining property of conditional expectation
# ------------------------------------------------------------------------------
print("\nQUESTION 5: Verifying Integral Property for all A in sigma(Y)")
print("-"*40)

verification_records = []
for A in sigma_Y:
    # Convert set representation to boolean mask
    mask = [om in A for om in omega]

    # Calculate integral (expectation contribution) over subset A
    integral_LHS = sum(df.loc[mask, 'E[X|Y]'] * df.loc[mask, 'Probability P(omega)'])
    integral_RHS = sum(df.loc[mask, 'Fraud Indicator (X)'] * df.loc[mask, 'Probability P(omega)'])

    match = np.isclose(integral_LHS, integral_RHS)
    verification_records.append({
        'Event A': str(A) if A else '{}',
        'Integral_A E[X|Y] dP': round(integral_LHS, 4),
        'Integral_A X dP': round(integral_RHS, 4),
        'Verified': match
    })

print(pd.DataFrame(verification_records).to_string(index=False))


# ------------------------------------------------------------------------------
# Q6 & Q7. Conditional PMF and Identity Verification
# ------------------------------------------------------------------------------
print("\nQUESTION 6 & 7: Conditional PMF & PMF Expectation Identity")
print("-"*40)

# Matrix setup for conditional PMF: p_{X|Y}(x | y)
# Rows = X (0, 1), Columns = Y (0, 1)
pmf_cond = np.zeros((2, 2))
# For Y = 0
pmf_cond[0, 0] = df[(df['Fraud Indicator (X)'] == 0) & (df['Risk Class (Y)'] == 0)]['Probability P(omega)'].sum() / p_y0
pmf_cond[1, 0] = df[(df['Fraud Indicator (X)'] == 1) & (df['Risk Class (Y)'] == 0)]['Probability P(omega)'].sum() / p_y0
# For Y = 1
pmf_cond[0, 1] = df[(df['Fraud Indicator (X)'] == 0) & (df['Risk Class (Y)'] == 1)]['Probability P(omega)'].sum() / p_y1
pmf_cond[1, 1] = df[(df['Fraud Indicator (X)'] == 1) & (df['Risk Class (Y)'] == 1)]['Probability P(omega)'].sum() / p_y1

print("Conditional PMF Matrix p_{X|Y}(x | y):")
print(f"  p_{{X|Y}}(0 | 0) = {pmf_cond[0,0]:.4f} (5/6) | p_{{X|Y}}(1 | 0) = {pmf_cond[1,0]:.4f} (1/6)")
print(f"  p_{{X|Y}}(0 | 1) = {pmf_cond[0,1]:.4f} (1/2) | p_{{X|Y}}(1 | 1) = {pmf_cond[1,1]:.4f} (1/2)")

print("\nVerifying Identity sum( x * p_{X|Y}(x|y) ):")
calc_identity_y0 = 0 * pmf_cond[0,0] + 1 * pmf_cond[1,0]
calc_identity_y1 = 0 * pmf_cond[0,1] + 1 * pmf_cond[1,1]
print(f"  For y=0: Calculated value = {calc_identity_y0:.4f} vs E[X|Y=0] = {E_X_Y0:.4f}")
print(f"  For y=1: Calculated value = {calc_identity_y1:.4f} vs E[X|Y=1] = {E_X_Y1:.4f}")


# ------------------------------------------------------------------------------
# Q8. Calculate the Mean Square Error (MSE) of predictor E[X|Y]
# ------------------------------------------------------------------------------
print("\nQUESTION 8: MSE of Optimal Predictor E[X|Y]")
print("-"*40)
# MSE = E[(X - E[X|Y])^2]
df['Sq_Error_Optimal'] = (df['Fraud Indicator (X)'] - df['E[X|Y]'])**2
mse_optimal = sum(df['Sq_Error_Optimal'] * df['Probability P(omega)'])
print(f"MSE of predictor E[X|Y] = {mse_optimal:.4f}")


# ------------------------------------------------------------------------------
# Q9. MSE using global average fraud rate E[X] and Law of Total Variance
# ------------------------------------------------------------------------------
print("\nQUESTION 9: Constant Predictor MSE vs Law of Total Variance")
print("-"*40)
# Global Expectation E[X]
E_X = sum(df['Fraud Indicator (X)'] * df['Probability P(omega)'])
print(f"Global Average Fraud Rate E[X] = {E_X:.2f}")

df['Sq_Error_Constant'] = (df['Fraud Indicator (X)'] - E_X)**2
mse_constant = sum(df['Sq_Error_Constant'] * df['Probability P(omega)'])
print(f"MSE of Baseline Constant Predictor E[X] = Var(X) = {mse_constant:.4f}")

# Law of Total Variance calculations: Var(X) = E[Var(X|Y)] + Var(E[X|Y])
# Var(X|Y) for Y=0
var_X_y0 = ((0 - E_X_Y0)**2 * pmf_cond[0,0]) + ((1 - E_X_Y0)**2 * pmf_cond[1,0])
# Var(X|Y) for Y=1
var_X_y1 = ((0 - E_X_Y1)**2 * pmf_cond[0,1]) + ((1 - E_X_Y1)**2 * pmf_cond[1,1])

E_Var_X_Y = (var_X_y0 * p_y0) + (var_X_y1 * p_y1)
Var_E_X_Y = (p_y0 * (E_X_Y0 - E_X)**2) + (p_y1 * (E_X_Y1 - E_X)**2)

print(f"\nBreakdown from Law of Total Variance:")
print(f"  E[Var(X|Y)] (Unexplained Variance / Optimal MSE) = {E_Var_X_Y:.4f}")
print(f"  Var(E[X|Y]) (Explained Variance by Risk Class)   = {Var_E_X_Y:.4f}")
print(f"  Total Var(X) = E[Var(X|Y)] + Var(E[X|Y])          = {E_Var_X_Y + Var_E_X_Y:.4f}")


# ------------------------------------------------------------------------------
# Q10. Information Theory Calculations
# ------------------------------------------------------------------------------
print("\nQUESTION 10: Information Theoretic Metrics")
print("-"*40)

# Marginal Distribution values for X
p_x1 = df[df['Fraud Indicator (X)'] == 1]['Probability P(omega)'].sum()
p_x0 = 1 - p_x1

# a) Entropy H(X)
H_X = - (p_x0 * np.log2(p_x0) + p_x1 * np.log2(p_x1))
print(f"a) Entropy H(X) (Initial Uncertainty) = {H_X:.4f} bits")

# b) Conditional Entropy H(X|Y) = sum( p(y) * H(X|Y=y) )
H_X_y0 = - (pmf_cond[0,0]*np.log2(pmf_cond[0,0]) + pmf_cond[1,0]*np.log2(pmf_cond[1,0]))
H_X_y1 = - (pmf_cond[0,1]*np.log2(pmf_cond[0,1]) + pmf_cond[1,1]*np.log2(pmf_cond[1,1]))
H_X_Y = (p_y0 * H_X_y0) + (p_y1 * H_X_y1)
print(f"b) Conditional Entropy H(X|Y) (Remaining Uncertainty) = {H_X_Y:.4f} bits")

# c) Mutual Information I(X;Y)
I_X_Y = H_X - H_X_Y
print(f"c) Mutual Information I(X;Y) (Information Shared) = {I_X_Y:.4f} bits")

# d) Percentage Uncertainty reduction
percentage_reduction = (I_X_Y / H_X) * 100
print(f"d) Percentage of uncertainty removed = {percentage_reduction:.2f}%")


# ------------------------------------------------------------------------------
# Visual Representation of Variance Decomposition
# ------------------------------------------------------------------------------
fig, ax = plt.subplots(figsize=(6, 4))
metrics = ['Total Variance\n(Baseline MSE)', 'Unexplained Var\n(Optimal MSE)', 'Explained Var\n(Var(E[X|Y]))']
values = [mse_constant, E_Var_X_Y, Var_E_X_Y]
colors = ['#e74c3c', '#3498db', '#2ecc71']

ax.bar(metrics, values, color=colors, edgecolor='black', width=0.5)
ax.set_ylabel('Variance / Error Value')
ax.set_title('Variance Decomposition: Law of Total Variance View')
for i, v in enumerate(values):
    ax.text(i, v + 0.01, f"{v:.4f}", ha='center', weight='bold')
plt.tight_layout()
plt.show()

print("\n" + "="*80)
print("LAB WORK COMPLETION SUCCESSFUL - NO WARNINGS DETECTED")
print("="*80)
```

```python
import numpy as np
import scipy.stats as stats

print("================================================================================")
print("             Q2) DERIVATION OF THE CONDITIONAL GAUSSIAN DENSITY")
print("================================================================================")
print("""
f_{X|Y}(x | y) = 1 / (sqrt((2*pi)^n * |S_XX|)) * exp(-0.5 * (x - mu*)^T * inv(S_XX) * (x - mu*))

Where:
Sigma_{X|Y} = S_XX = Sigma_XX - Sigma_XY * inv(Sigma_YY) * Sigma_YX
mu* = mu_X + Sigma_XY * inv(Sigma_YY) * (y - mu_Y)
""")

# ==============================================================================
# NUMERICAL VERIFICATION (Mathematical Ground Truth Comparison)
# ==============================================================================
print("================================================================================")
print("                    MATHEMATICAL INTEGRITY VERIFICATION")
print("================================================================================")

# Dimensions: n dimensions for X, m dimensions for Y
n = 2
m = 2

# Set random seed for perfect reproducibility
np.random.seed(42)

# Define Mean Vectors (mu_X and mu_Y)
mu_X = np.array([1.0, 2.0])
mu_Y = np.array([-1.0, 0.5])
mu_joint = np.concatenate([mu_X, mu_Y])

# Create a guaranteed symmetric, positive-definite joint Covariance Matrix (Sigma)
A = np.random.randn(n + m, n + m)
Sigma_joint = np.dot(A, A.T) + np.eye(n + m) * 0.1

# Partition the Covariance Matrix components
Sigma_XX = Sigma_joint[0:n, 0:n]
Sigma_XY = Sigma_joint[0:n, n:n+m]
Sigma_YX = Sigma_joint[n:n+m, 0:n]
Sigma_YY = Sigma_joint[n:n+m, n:n+m]

# Define an arbitrary given evaluation vector target
x_test = np.array([1.5, 1.8])
y_obs = np.array([0.0, 1.0])
z_joint = np.concatenate([x_test, y_obs])

# --- METHOD 1: YOUR RECONSTRUCTED CONDITIONAL FORMULA ---
Sigma_YY_inv = np.linalg.inv(Sigma_YY)

# 1. Calculated Parameters
Sigma_X_given_Y_analytical = Sigma_XX - np.dot(Sigma_XY, np.dot(Sigma_YY_inv, Sigma_YX))
mu_star_analytical = mu_X + np.dot(Sigma_XY, np.dot(Sigma_YY_inv, (y_obs - mu_Y)))

# 2. Probability Density Evaluation at x_test using your derived formula parameters
conditional_dist = stats.multivariate_normal(mean=mu_star_analytical, cov=Sigma_X_given_Y_analytical)
pdf_from_conditional_formula = conditional_dist.pdf(x_test)

# --- METHOD 2: RIGOROUS PROBABILITY THEORY PROOF DEFINITION ---
# f(x|y) = f(x,y) / f(y)
joint_dist = stats.multivariate_normal(mean=mu_joint, cov=Sigma_joint)
marginal_Y_dist = stats.multivariate_normal(mean=mu_Y, cov=Sigma_YY)

pdf_joint = joint_dist.pdf(z_joint)
pdf_marginal_Y = marginal_Y_dist.pdf(y_obs)

pdf_from_theoretical_definition = pdf_joint / pdf_marginal_Y

# --- OUPUT PROFILE COMPILATION ---
print("--- CALCULATED ANALYTICAL PROFILE ---")
print("Conditional Mean Vector (mu*):\n", mu_star_analytical)
print("\nConditional Covariance Matrix (Sigma_{X|Y}):\n", Sigma_X_given_Y_analytical)
print("\n--------------------------------------------------------------------------------")
print("--- COMPARATIVE DENSITY EVALUATION AT TEST POINT ---")
print(f"Density via Derived Parameters Formula : {pdf_from_conditional_formula:.12f}")
print(f"Density via Direct Definition Quotient : {pdf_from_theoretical_definition:.12f}")
print("\n--------------------------------------------------------------------------------")

# Quantify exact machine-precision absolute error
absolute_density_error = np.abs(pdf_from_conditional_formula - pdf_from_theoretical_definition)
print(f"Absolute Mathematical Profile Error: {absolute_density_error:.12e}")

if absolute_density_error < 1e-10:
    print("\n[STATUS: SUCCESS] The analytical density formulas are mathematically infallible and match to machine precision.")
else:
    print("\n[STATUS: WARNING] Numerical mismatch noted.")
```

```python
# @title Continuous Fault Monitoring of a Spindle Motor: Analysis and Simulation
# @markdown Run this cell to compute the analytical distributions, create the Minimum Mean Squared Error (MMSE) estimator, and visualize the system's behavior.

import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# ==============================================================================
# 1. SETUP & THEORETICAL DERIVATIONS
# ==============================================================================
"""
Problem Setup:
- Hidden Temperature:  X ~ N(0, 1)    => Mean(X) = 0, Var(X) = 1
- Mechanical Load:     Q ~ N(0, 4)    => Mean(Q) = 0, Var(Q) = 4
- Measurement (Power): Y = X + Q      (X and Q are independent)

Theoretical Properties:
1. Mean of Y:
   E[Y] = E[X] + E[Q] = 0 + 0 = 0

2. Variance of Y:
   Var(Y) = Var(X) + Var(Q) = 1 + 4 = 5

3. Covariance of X and Y:
   Cov(X, Y) = Cov(X, X + Q) = Var(X) + Cov(X, Q) = 1 + 0 = 1

4. Joint Distribution of [X, Y]^T:
   Bivariate Normal with Mean vector [0, 0]^T and Covariance Matrix:
   [[Var(X), Cov(X,Y)],
    [Cov(X,Y), Var(Y)]] = [[1, 1],
                           [1, 5]]

5. Posterior Distribution (X given Y = y):
   Using the properties of Conditional Gaussians:
   X | Y=y ~ N( mu_post, sigma_post^2 )
   where:
     mu_post    = E[X] + Cov(X,Y)/Var(Y) * (y - E[Y]) = 0 + (1/5) * (y - 0) = 0.2 * y
     sigma_post^2 = Var(X) - (Cov(X,Y)^2 / Var(Y))     = 1 - (1^2 / 5)       = 0.8
"""

# Define parameters explicitly
mu_X, var_X = 0, 1
mu_Q, var_Q = 0, 4

mu_Y = mu_X + mu_Q
var_Y = var_X + var_Q
cov_XY = var_X

print("="*60)
print("             THEORETICAL ANALYSIS SOLUTIONS")
print("="*60)
print(f"1. Total Spindle Power Distribution:  Y ~ N(mean={mu_Y}, variance={var_Y})")
print(f"2. Covariance between X and Y:        Cov(X, Y) = {cov_XY}")
print(f"3. Joint Distribution Matrix [X, Y]:  Mean = [0, 0]")
print(f"                                      Cov  = [[{var_X}, {cov_XY}],")
print(f"                                              [{cov_XY}, {var_Y}]]")
print(f"4. MMSE Estimator Function:           E[X|Y=y] = {cov_XY/var_Y} * Y = 0.2 * Y")
print(f"5. Posterior Error Variance:          Var(X|Y) = {var_X - (cov_XY**2/var_Y)}")
print("="*60 + "\n")


# ==============================================================================
# 2. NUMERICAL SIMULATION
# ==============================================================================
np.random.seed(42) # Set seed for reproducibility
num_samples = 100000

# Simulate the physical processes
X_samples = np.random.normal(mu_X, np.sqrt(var_X), num_samples)
Q_samples = np.random.normal(mu_Q, np.sqrt(var_Q), num_samples)
Y_samples = X_samples + Q_samples

# Apply MMSE Estimator to guess hidden temperature X purely from observed power Y
X_estimated = (cov_XY / var_Y) * Y_samples
estimation_errors = X_samples - X_estimated

# Compute Empirical (Simulated) Metrics
sim_mu_Y = np.mean(Y_samples)
sim_var_Y = np.var(Y_samples)
sim_cov_XY = np.cov(X_samples, Y_samples)[0, 1]
sim_mse = np.mean(estimation_errors**2)

print("="*60)
print("             SIMULATION VERIFICATION METRICS")
print("="*60)
print(f"Simulated Mean of Y:              {sim_mu_Y:.4f}  (Expected: {mu_Y})")
print(f"Simulated Variance of Y:          {sim_var_Y:.4f}  (Expected: {var_Y})")
print(f"Simulated Covariance(X,Y):        {sim_cov_XY:.4f}  (Expected: {cov_XY})")
print(f"Simulated Mean Squared Error:     {sim_mse:.4f}  (Expected/Minimum: {var_X - (cov_XY**2/var_Y)})")
print("="*60 + "\n")


# ==============================================================================
# 3. ADVANCED VISUALIZATION
# ==============================================================================
fig, axs = plt.subplots(1, 2, figsize=(15, 6))

# Plot 1: Joint Distribution Density & Estimator Line
hb = axs[0].hexbin(Y_samples, X_samples, gridsize=40, cmap='YlOrRd', mincnt=1)
y_range = np.linspace(np.min(Y_samples), np.max(Y_samples), 100)
x_predicted = (cov_XY / var_Y) * y_range
axs[0].plot(y_range, x_predicted, color='blue', linestyle='--', linewidth=2.5,
            label=r'MMSE Estimate: $\hat{X} = 0.2Y$')

axs[0].set_title("Joint Distribution & Temperature Estimation Line", fontsize=12, fontweight='bold')
axs[0].set_xlabel("Observed Total Spindle Power (Y)", fontsize=11)
axs[0].set_ylabel("Hidden Internal Temperature (X)", fontsize=11)
axs[0].grid(True, alpha=0.3)
axs[0].legend(loc='upper left')
cb = fig.colorbar(hb, ax=axs[0], label='Sample Density Count')

# Plot 2: Prior vs. Posterior Uncertainties (Given an arbitrary anomalous reading Y = 3.5)
specific_y = 3.5
post_mean = (cov_XY / var_Y) * specific_y
post_var = var_X - (cov_XY**2 / var_Y)
post_std = np.sqrt(post_var)

x_axis = np.linspace(-4, 4, 500)
prior_pdf = stats.norm.pdf(x_axis, mu_X, np.sqrt(var_X))
posterior_pdf = stats.norm.pdf(x_axis, post_mean, post_std)

axs[1].plot(x_axis, prior_pdf, 'g-', label=r'Prior: $X \sim \mathcal{N}(0,1)$', linewidth=2)
axs[1].plot(x_axis, posterior_pdf, 'b-', label=r'Posterior given $Y=3.5$: $X|Y \sim \mathcal{N}(0.7, 0.8)$', linewidth=2)
axs[1].axvline(post_mean, color='red', linestyle=':', label=r'Best Estimate ($\hat{x} = 0.7$)')

axs[1].set_title(f"Uncertainty Reduction via Observation (Example Y = {specific_y})", fontsize=12, fontweight='bold')
axs[1].set_xlabel("Internal Temperature Deviation (X)", fontsize=11)
axs[1].set_ylabel("Probability Density", fontsize=11)
axs[1].grid(True, alpha=0.3)
axs[1].legend(loc='upper left')

plt.tight_layout()
plt.show()
```
